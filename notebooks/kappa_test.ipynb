## Import data

```python
import pandas as pd

annotator_1 = pd.read_csv("../data/blind_labelling_annotator_1.csv")
annotator_2 = pd.read_csv("../data/blind_labelling_annotator_2.csv")

print("Annotator 1 = Ghif")
print("Annotator 2 = Richard")

display(annotator_1.head())
display(annotator_2.head())
```

## Data sanity

```python
print("Same number of rows:", len(annotator_1) == len(annotator_2))
print("Same case_id order:", (annotator_1["case_id"] == annotator_2["case_id"]).all())
print("Duplicate case_id annotator 1:", annotator_1["case_id"].duplicated().sum())
print("Duplicate case_id annotator 2:", annotator_2["case_id"].duplicated().sum())
print("Missing case_id annotator 1:", annotator_1["case_id"].isna().sum())
print("Missing case_id annotator 2:", annotator_2["case_id"].isna().sum())
```

## Kappa Test

```python
from sklearn.metrics import cohen_kappa_score

# Toxicity
toxicity_pair = pd.DataFrame({
    "annotator_1": annotator_1["toxicity_clean"],
    "annotator_2": annotator_2["toxicity_clean"]
}).dropna()

toxicity_agreement = (toxicity_pair["annotator_1"] == toxicity_pair["annotator_2"]).mean()
toxicity_kappa = cohen_kappa_score(
    toxicity_pair["annotator_1"],
    toxicity_pair["annotator_2"]
)

# Irony
irony_pair = pd.DataFrame({
    "annotator_1": annotator_1["irony_clean"],
    "annotator_2": annotator_2["irony_clean"]
}).dropna()

irony_agreement = (irony_pair["annotator_1"] == irony_pair["annotator_2"]).mean()
irony_kappa = cohen_kappa_score(
    irony_pair["annotator_1"],
    irony_pair["annotator_2"]
)

print("Toxicity agreement:", toxicity_agreement)
print("Toxicity kappa:", toxicity_kappa)
print("Irony agreement:", irony_agreement)
print("Irony kappa:", irony_kappa)
```
